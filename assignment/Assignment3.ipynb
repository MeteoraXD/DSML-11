```python
# List of Students
Students = [ "Ram" , "Sita" , "Gita" , "Hari" , "Shyam" , "Rita" ]
# Marks of Studens 
Marks = [ 85 , 42 , 78 , 91 , 35 , 67 ]

```

```python
# Display Students Serial Numbers starting from 1
for index, student in enumerate(Students, start=1):
    print(f"{index}. {student}")  
```

```python
# Combime the sturdents names and marks into a single collection containing students records
Student_Records = dict(zip(Students, Marks))
# Display the Student Records
print("\nEvery Student Records:" , Student_Records)
```

```python
# Add Grade System to the Student Records.
def get_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B+"
    elif marks >= 60:
        return "B"
    elif marks >= 50:
        return "C+"
    elif marks >= 40:
        return "C"
    elif marks >= 30:
        return "D"
    else:
        return "F"

results = {}
for key, value in Student_Records.items():
    results[key] = [value,get_grade(value)]

for key, value in results.items():
    print(f"Name : {key}  Marks Obtained : {value[0]}  Grade : {value[1]}")
     
```

```python
# Create a list containing only the students who have passed the examination. (marks above 40 is pass)

Passed_Students = [student for student, marks in Student_Records.items() if marks > 40]
print("\n The Passed Students are 
:" , Passed_Students)
```

```python
# Calculate the total marks obtained by all students combined.

total_marks = sum(Student_Records.values())
print("\n The Total Marks Obtained by All Students:" , total_marks)
```

```python
# Find the highest mark among all students without using the built-in max() function.

highest_mark = Student_Records[Students[0]]
for marks in Student_Records.values():
    if marks > highest_mark:
        highest_mark = marks
print("\nHighest Mark Obtained by Any Student: " , highest_mark)
```

```python
# Upper converts every letter in string to Uppercase.
my_string = 'sanjay'
print(my_string.upper())

```

```python
# Capitalize converts first letter of string to Uppercase.
my_string2 = 'sanjay'
print(my_string2.capitalize())
```

```python
# Title capitalizes first letter of every word in a string.
my_string3 = " i am learning python"
print(my_string3.title())
```

```python
# Strip removes whitespace from both ends
my_String4 = "        Hello World      "
print(my_String4.strip()) 
```

```python
# rstrip removes whitespace from right side 
my_string5 = "        Hello World      "
print(my_string5.rstrip())
```

```python
# lstrip removes whitespace from left side
my_String6 = "       Hello World   "
print(my_String6.lstrip())
```

```python
# replace() method replaces parts of a string.

my_string7 = "I am a web developer"
print(my_string7.replace("web" , "mobile"))
```

```python
# Join method joins elements of a list with splitter
my_string8 = [ " web" , "developer"]
joined_string = "".join(my_string8)
print(joined_string)
```

```python
# count() method counts how many time a substring has occured
my_string9 = "I am learning Python. I know Javascript"
print(my_string9.count('I'))
```

```python
# startswith() method checks if the string starts with given arguments.
my_string10 = "I am using Google Colab."
print(my_string10.startswith('I'))
```

```python
# Take input from user
name = input("Enter your name: ")
username = f"my name is {name}"

# replace() method replaces the existing string with another given string.
text_Replacer = username.replace(name , "pahari")
print(text_Replacer)
```

```python
# convert existing string to upper case
print(text_Replacer.upper())
```

```python
# find() method returns index of first occurence 

print(text_Replacer.find("s"))
```

```python
# index method is similar to find()
print(text_Replacer.index('y'))
```

```python
# len() method shows the length of the string.
print(len(text_Replacer))
```

```python
# isalnum checks if all chracters are alphanumeric
print(text_Replacer.isalnum())  
```

```python
# isalpha() checks if all characters are alphabetic
print(text_Replacer.isalpha())
```

```python
# isdigit() checks if all characters are  numeric
print(text_Replacer.isdigit())
```

```python
# islower() checks if all characters are lower case
print(text_Replacer.islower())
```

```python
# isupper() checks if all characters are lower case
print(text_Replacer.isupper())
```

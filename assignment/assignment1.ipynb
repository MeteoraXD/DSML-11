```python
# 1.Upper converts every letter in string to Uppercase.
my_string = 'sanjay'
print(my_string.upper())

```

```python
# 2.Capitalize converts first letter of string to Uppercase.
my_string2 = 'sanjay'
print(my_string2.capitalize())
```

```python
# Title capitalizes first letter of every word in a string.
my_string3 = " i am learning python"
print(my_string3.title())
```

```python
# Strip removes whitespace from both ends
my_String4 = "        Hello World      "
print(my_String4.strip()) 
```

```python
# rstrip removes whitespace from right side 
my_string5 = "        Hello World      "
print(my_string5.rstrip())
```

```python
# lstrip removes whitespace from left side
my_String6 = "       Hello World   "
print(my_String6.lstrip())
```

```python
# replace() method replaces parts of a string.

my_string7 = "I am a web developer"
print(my_string7.replace("web" , "mobile"))
```

```python
# Join method joins elements of a list with splitter
my_string8 = [ " web" , "developer"]
x = "".join(my_string8)
print(x)
```

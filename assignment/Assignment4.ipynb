# Student Grade Management System
Create a Python class named Student.
Requirements:
- Create a class attribute named school.
- Use the __init__() method to take:
    -Student name
    -Student marks
- Create an instance method named check_grade() that displays:
    - A Grade: Marks ≥ 80
    - B Grade: Marks ≥ 60 and < 80
    - C Grade: Marks ≥ 40 and < 60
    - Fail: Marks < 40
- Create a class method to display the school name.
- Create a static method to display a welcome message.
- Create an object and display the student's grade

```python
class Student:
    school = "DeMuerte School"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def check_grade(self):
        if self.marks >= 80:
            print(f"{self.name}: Grade A ")
        elif self.marks >= 60:
            print(f"{self.name}: Grade B ")
        elif self.marks >= 40:
            print(f"{self.name}: Grade C ")
        elif self.marks > 100 & self.marks < 0 :
            print("Invalid marks provided.") 
        else:
            print(f"{self.name}: Fail ")

    @classmethod
    def show_school(cls):
        print(f"School: {cls.school}")

    @staticmethod
    def welcome():
        print("Welcome to the Student Grade System!")





```

```python
Student.welcome()
Student.show_school()
```

```python

s1 = Student("Ramesh", 85)
s1.check_grade()
```

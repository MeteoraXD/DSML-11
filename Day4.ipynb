```python
# deep copy and shadow copy
list99 = [ 1,2,3,4]
list100 = list99
print(list99)
print(list100)
```

```python
#shadow copy
list101 = list99.copy()
list99.append(8)
list101
print(list99)
```

```python
lst11 = [[1,2,3] , [ 4,5,6]]
list12 = lst11.copy()
lst11[1][1] = 100
print(lst11)
print(list12)
```

```python
import copy
lst13 = [[1,2,3] , [ 4,5,6]]
lst14 = copy.deepcopy(lst13)
lst13[1][2]  = 100
print(lst13)
print(lst14)
```

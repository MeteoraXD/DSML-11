```python
# no parameter , no return type
def greet():
    print("Hello How are you?")
    print("Have a great day")

greet()
```

```python
# has parameter no return type
def mathematical_operations(a,b):
    print(a+b)
    print(a-b)
    print(a/b)
    print(a*b)
    print(a%b)

mathematical_operations(5,6)
```

```python
def vowel_count(a):
    vowels = 'aeiouAEIOU'
    count = 0

    for char in a:
        if char in vowels:
            count += 1

    return count

a = input('Enter a String: ')

print(vowel_count(a))
```

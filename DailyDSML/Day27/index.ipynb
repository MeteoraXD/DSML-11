```python
import pandas as pd

a = [ '1' , '2 ' , '3' , '4 ' , '5']
res = pd.Series(a , index=['a' , 'b' , 'c' , 'd' , 'e' , ] )
res
```

```python
dic = {
    id: 1,
    'Name': 'Sanjay',
    'State' : 'Bagmati',
    'Age': 25
}

res_dict = pd.Series(dic)
df = pd.DataFrame(dic , index=[1])
res_dict
df
```

```python
# rows and columns ko combination lai Dataframe vanxa
# simple data frame

data = [['python' , 'java' , 'javascript' , 'C++']]
df = pd.DataFrame(data, index=['1'], columns=['Lang1', 'Lang2', 'Lang3', 'Lang4'])
print(df)
```

```python

data = ['python' , 'java' , 'javascript' , 'C++']
df = pd.DataFrame(data , index=['1' , '2' , '3' , '4' ] , columns=['Language'])
df
```

```python
# dictionary data frame

dict_data = {
    'Name': ['Sanjay' , 'Shishir' , 'Sujan'],
    'marks' : [89 , 90 , 92]
}
df = pd.DataFrame(dict_data)
df
```

```python
# rows and columns ko combination lai Dataframe vanxa
# simple data frame

data = [['python' , 'java' , 'javascript' , 'C++']]
df = pd.DataFrame(data, index=['1'], columns=['Lang1', 'Lang2', 'Lang3', 'Lang4'])
print(df)
```

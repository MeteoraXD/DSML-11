```python
def sum(a,b):
    '''Adding two numbers
    '''
    return(a+b)

print(sum.__doc__)
```

```python
help(sum)
```

```python
def area_of_rectangle(l,b):
    '''Area of the rectangle has length(l) and breadth(b)
    '''
    return(l*b)

print(area_of_rectangle.__doc__)
print(area_of_rectangle(4,6))
```

```python
help(area_of_rectangle)
```

```python
def data3(n, *args , **kwargs):
    print('The number of pizza is:',n)
    print('Types of Pizza customer likes ',args)
    print('User Address' ,kwargs)

data3(2,'Ham','Bacon',name='John',address='123 Street, City')  
```

```python
def pizza_order(n, *args, **kwargs):
    print("Number of pizzas:", n)
    print("Types of pizza:", args)
    print("Customer info:", kwargs)

n = int(input('Enter the number of pizzas: '))
lst_1 = []

for _ in range(n):
    types_of_pizza = input('What type of pizza would you like? ')
    lst_1.append(types_of_pizza)

name_1 = input('Enter your name: ')
address_1 = input('Enter your address: ')

pizza_order(n, lst_1, name=name_1, address=address_1)
```

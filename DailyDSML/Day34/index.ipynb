```python
import pandas as pd
df1 = pd.DataFrame({
    "acc": [1,2,3],
    "name": [0,9,8],
    "age":[3,5,7]
})
```

```python
df1
```

```python
df2 = pd.DataFrame({
    "acc": [1,24,33],
    "loan": [17,29,58],
})
```

```python
concaticated = pd.concat([df1 , df2 ],axis=0)
concaticated
```

```python
concaticated_col = pd.concat([df1 , df2 ],axis=1)
```

```python
concaticated_col
```

```python
# Merge ley 2 or more data frames

pd.merge(df1,df2 , how="inner", suffixes=("__left"  , "_right"))
```

```python
# join ley datafram join garauxa

df3 = pd.DataFrame({
    "key": [1,2,3],
    "name": [0,9,8],
    "age":[3,5,7]
})

df_3 = df3.set_index("key")
df_3
```

```python
joined_df = df1.join(df3, how="right", lsuffix="_x") 
```

```python
joined_df
```

```python
df2.join(df1,how="right",rsuffix="_2")
```

```python
stacked_df = df1.stack()
```

```python
stacked_df
```

```python
unstacked_df = stacked_df.unstack()
```

```python
unstacked_df
```

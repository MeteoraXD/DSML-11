```python
class Calc:
    def add(self, a,b):
        return(a+b)
    def add(self,a,b,c):    
        return(a+b+c)

           
```

```python
obj = Calc()
obj.add(2,3,4)
```

```python
class Calculator:
    def add(self, *args):
        return sum(args)
        
```

```python
obj_2 = Calculator
obj_2.add(1,2,3,4,5,6)
```

**Inheritance**

```python
# Simple Inheritance

class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def bark(self):
        print('Dog barks')
        
```

```python
d = Dog()
d.bark()
d.sound()
```

```python
# Multiple Inheritance

class Mammals:
    def fur(self):
        print("Mammal has fur")

class Herbivore:
    def eat(self):
        print("Herbivore only eat veg")

class Goat(Mammals, Herbivore):
    def identity(self):
        print('Goat has fur and are herbivore')


```

```python
g = Goat()
g.eat()
g.fur()
g.identity()
```

```python
# MultiLevel Inheritance

class Mammal:
    def blood(self):
        print('Warm Blooded')

class Dogs(Mammal):
    def bark(self):
        print("Dog barks")

class Labrador(Dogs):
    def lab(self):
        print('have fur')
```

```python
l1 = Labrador()
l1.bark()
l1.blood()
l1.lab()
```

```python
# Hierechial Inheritance

class Mother:
    def skill(self):
        print("Cooking")

class child1(Mother):
    def identity(self):
        print('child1')
class child2(Mother):
    def identity(self):
        print('child2')
```

```python
c1 = child1()
c2 = child2()
c1.identity()
c2.skill()
```

```python
# hybird inheritance

class Person:
    def details(self):
        print('Ram is 21 yrs old')

class Student(Person):
    def student_id(self):
        print('9812yt')

class Teacher(Person):
    def subject(self):
        print('Python')

class TeachingAssistance(Student,Teacher):
    def salary_department(self):
        print('IT and 55000 ')


```

```python
p = TeachingAssistance()
p.details()
p.student_id()
p.subject()
p.salary_department()
```

```python
class Person1:
    def __init__(self):
        self.name = str(input("Enter your name"))
        self.age = int(input("Enter your age"))

    def details(self):
        print(f"{self.name} is {self.age} yrs old")

class Student1(Person1):
    def student_id(self,std_id):
        self.std_id = std_id
        print(f"{self.name}'s student id is {self.std_id}")

class Teacher1(Person1):
    def subject(self,sub , teacher_name):
        self.teacher_name = teacher_name
        self.sub = sub
        print(f"{self.teacher_name} teachers {sub} to {self.name}")

class TeachingAssistance1(Student1,Teacher1):
    def faculty(self,dep,salary):
        self.dep = dep
        self.salary = salary
        print(f'The Teachhing Assitance is from {self.dep} faculty and earns {self.salary} per month')

```

```python
obwan = TeachingAssistance1()
obwan.details()
obwan.student_id('2111yht')
obwan.subject('C++' , "ramesh")
obwan.faculty('IT' , 30000)
```

```python
class OTTPSubscription:
    def __init__(self,sub_id,plan, total_payment):
        self.id = sub_id
        self.plan = plan
        self.total_payment = total_payment

    def subscription(self):
        print(f"Subscription with id no: {self.id} is subscribed to the {self.plan} ")

    def unsubribe(self):
        print(f"Subscriber with {self.id} unsubsribed to the {self.plan}")        
    
        
```

```python
class Premium_Sub(OTTPSubscription):
    def __init__(self ,sub_id,plan, total_payment, screens):
        super().__init__(sub_id,plan,total_payment)
        self.max_screens = screens

    def set_max_screen(self,screens):
        self.max_screens = screens
        print(f"Maximum screen set to {self.max_screens} in premium plan")


```

```python
obj = Premium_Sub(2322, "enterprise" , 30000, 5)

```

```python
obj.set_max_screen(5)
obj.subscription()
obj.unsubribe()
obj.plan
obj.total_payment
```

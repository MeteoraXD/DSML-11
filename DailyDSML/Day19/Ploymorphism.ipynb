```python
class PaymentGateaway:
    def process_payment(self,amount):
        pass

class CreditCard(PaymentGateaway):
    def process_payment(self,amount):
        return(f"Processing ${amount} via Credit Card Authentication")

class Paypal(PaymentGateaway):
    def process_payment(self,amount):
        return(f"Processing ${amount} via Paypal secure token")

# unified interae function

def execute_checkout(payment_method , amount):
    print(payment_method.process_payment(amount))
        
```

```python
card = CreditCard()
card.process_payment(300)
wallet = Paypal()
```

```python
execute_checkout(card,250)
execute_checkout(wallet,250)
```

```python
class Student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2

    def __add__(self,other): #Student.__add__(s1,s2)
        m1 = self.m1 + other.m1
        m2 = self.m2 + other.m2
        s3 = Student(m1,m2)
        return s3

    def __gt__(self,other):
        return self.m1 > other.m1
        
```

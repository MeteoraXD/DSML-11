```python
lst = []

def user_data():
    for i in range(5):
        a = int(input("Enter a number: "))
        lst.append(a)

    return lst

data = user_data()

file = open("index.txt", "w")
file.write(str(data))
file.close()
```

```python
LST_1 = [1,2,3,4,5]

def square(x):
    return x*x

lst_2 = map(square,LST_1)
print(list(lst_2))

```

```python
lst = [1,2,3,4,5,6,7,8,9 ]

def even_odd(n):
    if n%2 == 0 :
        return(n)

res = filter(even_odd ,lst)          
print(list(res))
```

```python
# from functools import reduce

def mux(a,b):
    return(a*b)

lst_1 = [1,2,3,4,5]    
resData = reduce(mux,lst_1)
print(resData)

```

```python
lst2 = [ 2,5,7,9,1]

def maximum_nmber(a,b):
    if a>b:
        return a
    return b    

resData2 = reduce(maximum_nmber,lst2)
print(resData2)


```

```python
import pandas as pd
titanic_df = pd.read_csv("titanic.csv")
```

```python
import matplotlib as plot
import matplotlib.pyplot as plt
titanic_df["Sex"].value_counts().plot.bar()

```

```python
titanic_df["Survived"].value_counts().plot(
    kind="bar",
    title="Count",
    figsize=(5,5),
    legend= True,
    yticks=list(range(0,600,50)),
    xlabel="Survived (0= No , 1 = Yes)",
    ylabel= "No. of Passenger",
    grid =True,
    colormap="bwr"
)
```

```python
pd.crosstab(titanic_df["Survived"], titanic_df["Sex"]).plot(
    kind="bar",
    title="Survival Count by Sex",
    figsize=(5,5),
    legend=True,
    yticks=list(range(0,600,50)),
    xlabel="Survived (0=No, 1=Yes)",
    ylabel="No. of Passengers",
    grid=True,
    colormap="Dark2"
)
```

```python
titanic_df["Survived"].value_counts().plot(
    kind="pie",
    title="Survival Distribution",
    figsize=(5,5),
    autopct="%1.1f%%",
    labels=["No","Yes"],
    colormap="bwr"
)
```

```python
pd.crosstab(titanic_df["Survived"], titanic_df["Sex"]).plot()
```

```python
grouped = titanic_df[["Survived","Embarked"]].groupby(by="Embarked")
grouped.count().plot(
    kind="pie",
    y="Survived"
)
```

```python
ct = pd.crosstab(titanic_df["Sex"], titanic_df["Survived"])

fig, axes = plt.subplots(1, 2, figsize=(10,5))

ct.loc["male"].plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["No","Yes"],
    title="Male Survival",
    colormap="bwr",
    ax=axes[0]
)

ct.loc["female"].plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["No","Yes"],
    title="Female Survival",
    colormap="bwr",
    ax=axes[1]
)

plt.tight_layout()
plt.show()
```

```python
titanic_df["Survived"].value_counts().plot.pie(
    autopct = "%1.1f%%",
    figsize=(6,6)
    )
plt.ylabel("Survived vs  Not Survived")
plt.show()
```

```python
grouped = titanic_df[["Age","Embarked"]].groupby(by="Embarked")
grouped.count().plot(
    kind="pie",
    y="Age",
        autopct = "%1.1f%%",

)
```

```python
pd.crosstab(titanic_df.Survived,titanic_df.Sex).plot(
    kind ='pie',
    fontsize = 10,
    subplots = True,
    # figmap=(6,8),
    colormap='Dark2'
    )
```

```python
pd.crosstab(titanic_df.Survived,titanic_df.Sex).plot(
    kind ='line',
    fontsize = 10,
    # figmap=(6,8),
    colormap='Dark2'
    )
```

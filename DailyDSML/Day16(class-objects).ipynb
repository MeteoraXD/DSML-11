```python
class Student:
    '''Student Information'''
    school = 'XYZ' # class attribute 
    def __init__(self , name,age):
        self.name = name # instance attribute
        self.age = age
    @classmethod
    def get_school(cls):
        print(cls.school)
s1 = Student("Sanjay",20)

print(s1.age)
print(s1.name)
s1.get_school()
# class attribute lina @classmethod chainxa
```

```python
class Calculator:
    '''Calculator'''
    @staticmethod
    def add(a,b):
        return a+b
    def sub(a,b):
        return a-b
    def div(a,b):
        return a/b        
    def mux(a,b):
        return a*b
sum = Calculator.add(5,6)
subt = Calculator.sub(5,6)
mul = Calculator.mux(5,6)
divide = Calculator.div(5,6)
print(sum)   
print(subt)   
print(mul)   
print(divide)   
```

```python
class Dog:
    count = 0 # class attribute

    def __init__(self,name, age=3):
        self.name = name
        self.age = age
        Dog.count += 1
    
    def bark(self):
        print(f"{self.name} has barked")

    @classmethod
    def get_count(cls):
        return cls.count

    @staticmethod
    def get_vet_ingo():
        return "Dr. Mahesh Uncle ko chora"


```

```python
Dog1 = Dog('Bruno')
Dog1.name
```

```python
Dog1.age
```

```python
Dog1.bark()
```

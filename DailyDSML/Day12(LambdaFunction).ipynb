```python
# function = lambda(keyword) arguments : expression

data = lambda a,b : a+b
data(1,2)
```

```python
data = lambda a: a*a
data(5)
```

```python
# Map function using lambda function
lst = [1,2,3,4,5]
list(map(lambda x: x**2, lst))
```

```python
# Filter function using lambda function
lst = [1,2,3,4,5]
list(filter(lambda x: x % 2 == 0, lst))
```

```python
# Reduce function using lambda function
from functools import reduce

lst = [1,2,3,4,5]
reduce(lambda x, y: x * y, lst)
```

```python
# Even or Odd using lambda function
user_input = int(input("Enter a number: "))
odd_or_even = lambda x: "Even" if x % 2 == 0 else "Odd"
result = odd_or_even(user_input)
print(result)
```

```python
students = [
    {"name": "Alice", "salary": 22000},
    {"name": "Ramesh", "salary": 20000},
    {"name": "Charlie", "salary": 23000}]

students.sort(reverse=True, key=lambda x: x["salary"])
print(students)


```

```python
students_1 = [
    ( "Ramesh", 22),
    ( "Alice", 22),
    ( "Charlie", 20)
]
students_1.sort(reverse=True, key=lambda x: x[0])
print(students_1)
```

```python
cars = ['VW','Audi','Mercedes' , 'Mitshubishi']
cars.sort(reverse=False, key=lambda x: len(x))
print(cars)
```

```python
lst1 = [ 1,2,3,4,5]
res = [ val for val in lst1 if val % 2 == 0]
print(res)
```

```python
lst2 = [ 'welcome' , 'to' , 'pethon' , 'programming']

for word in lst2:
    if 'e' in word :
        print(word.upper())
    else:
        print(word)

```

```python
# {key : value for key, value in iterable} Dictionary comprehension

{i : i**2 for i in range(1,10) if i % 2 == 0}
```

```python
list8 = [1,2,3,4,5]

{num : num**2 for num in list8 if num % 2 == 0}
```

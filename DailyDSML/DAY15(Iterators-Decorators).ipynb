```python
lst = [ 1,2,3,4]
for i in lst:
    print(i)
```

```python
# iterator
num = [ 1,2,3,4]
it = iter(num)
print(next(it))
# shows the next element in the list step by step
```

```python
def square(n):
    for i in range(n):
        return i**2
square(3)        
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
a = square(3)
print(next(a))
```

```python
print(next(a))
```

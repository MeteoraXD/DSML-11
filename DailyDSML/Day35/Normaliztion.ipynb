```python
import pandas as pd
iris_df = pd.read_csv("Iris.csv",index_col="Id")
```

```python
iris_df.head()
```

```python
iris_df.describe()
```

```python
# Handlingg Data Normalization

min_length = iris_df.SepalLengthCm.min()
max_length = iris_df.SepalLengthCm.max()
```

```python
normal_length = (iris_df.SepalLengthCm)
```

```python
normalized  = (normal_length - min_length)/(max_length--min_length)
```

```python
normalized
```

```python
min_length_1 = iris_df.PetalLengthCm.min()
max_length_1 = iris_df.PetalLengthCm.max()
```

```python
normal_length_1 = (iris_df.PetalWidthCm)
```

```python
normalized_1  = (normal_length_1 - min_length_1)/(max_length_1-min_length_1)
```

```python
normalized_1
```

```python
data = {
    "Age": [25, 30, 35, 43 , 32,54],
}
```

```python
df
```

```python
df_min = df.Age.min()
df_max = df.Age.max()
df_normal = df.Age
normal_ages = (df_normal - df_min)/(df_max - df_min)
normal_ages.head()
```

```python
normal_ages.plot.kde()
```

```python
from sklearn.preprocessing import MinMaxScaler , StandardScaler
```

```python
minmax_scalar = MinMaxScaler(feature_range=(0,1))
```

```python
norm_data = minmax_scalar.fit_transform(iris_df.drop("Species", axis=1))
```

```python
norm_data
```

```python
pd.DataFrame(norm_data,columns=['SepalLengthCm','SepalWidthCm','PetalWidthCm','PetalLengthCm'])
```

```python
x= iris_df.SepalLengthCm
mean_df = iris_df.SepalLengthCm.mean()
std_df = iris_df.SepalLengthCm.std()

z_standard = (x-mean_df)/std_df
```

```python
z_standard
```

```python
round(z_standard.describe(),5)
```

```python
z_standard.plot.kde()
```

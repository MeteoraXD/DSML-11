```python
# Recusrsion
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))  
```

```python
# Fabionacci series
lst = []
def fabionacci_series(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        lst = fabionacci_series(n - 1)
        lst.append(lst[-1] + lst[-2])
        return lst

print(fabionacci_series(10))        
```

```python
# Function passed as arguments
def sum(a, b):
    return a+ b

def calculate(a, b, operation):
    return operation(a, b)
result = calculate(5, 3, sum)
print(result)  



```

```python
# Return function as a value

def greeting(name):
    def message():
        return f"Hello, {name}!"
    return message
greet = greeting("Sanjay")
print(greet())
```

```python
def my_Wrapper(func):
    def wrapper():
        print("Inside Wrapper.")
        func()
    return wrapper
```

```python
@my_Wrapper
def wrapper():
    print("Hello Wrapper.")
wrapper()
```

```python
def greetings(func):
    def wrapper(*args, **kwargs):
        # print(f"Calling {func.__name__}? ")
        return func(*args, **kwargs)
    return wrapper
```

```python
@greetings
def greet(name, age):
    print(f"Hello {name}.You are {age} years old.")

greet("Ram", 25)
```

```python
def greeting_1(func):
    def wrapper(*args, **kwargs):
        print("Before Function Execution.")
        result = func(*args, **kwargs)
        print("After Function Execution.")
        return result
    return wrapper
```

```python
@greeting_1
def greet(name):
    print(f"Hello {name}.")

greet("Ram")    
```

 - Method Overriding


```python
# Method Overriding

class Animal:
    def sound(self):
        print("Makes a soung")

class Dog(Animal):
    def sound(self):
        print("Barks")

class Cat(Animal):
    def sound(self):
        print("Meows")
```

```python
obj = Dog()
obj.sound()
```

```python
class Animal:
    def show_details(self):
        print("I am from Animal")

class Dog(Animal):
    def show_details(self):
        super().show_details()
        print('I am dog')

class Cat(Animal):
    def show_details(self):
        super().show_details()
        print('I am dog')
```

```python
obj1 = Dog()
obj1.show_details()
```

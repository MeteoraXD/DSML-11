# Encapsulation

Create a bank account class with:
- Account holder name is public
- Protected account type
- Private Pin and balance
- Create a Saving Account class that adds 5% interest to the balance using methods

```python
class BankAccount:
    def __init__(self,owner,account_type,pin,balance):
        self.owner = owner
        self._account_type = account_type
        self.__pin = pin
        self.__balance = balance
    def get_balance(self,pin):
        if pin == self.__pin:
            return(self.__balance)
        else:
            return("Invalid Pin entered.")    

    def deposit(self,amount):
        if amount >= 0:
            self.__balance += amount
            print(f"{amount} is deposited into our account.Your New balance is {self.__balance}")
        else:
            print("Invalid Amount entered")
    def check_balance(self):
        print(f"The Balance is  {self.__balance}")
        return self.__balance

    def account_type(self):
        print(f"The account type is {self._account_type}")  
    def pin(self):
        print(f"The account type is {self.__pin}")   
```

```python
class SavingAccount(BankAccount):
    def interest(self, rate=5):
        balance = self.check_balance()
        interest = (balance * rate) / 100
        balance += interest
        print(f"Account interest : {interest}")
        return balance
```

```python
account_type = str(input("enter the type of account you want"))
name = str(input("enter your name"))
initial_balance = int(input("enter the balance you want to deposit"))
pin = int(input("Enter your PIN"))

obj = SavingAccount(name ,account_type,pin,initial_balance )
obj.get_balance(pin)
obj.check_balance()
obj.interest()
obj.account_type()
obj.pin()
```

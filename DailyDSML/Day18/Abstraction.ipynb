# Abstraction

Create an abstract BankAccount class with abstract methods deposit() and withdraw(). Implement them in a saving account class.

```python
from abc import ABC , abstractmethod

class BankAccount(ABC):
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance

    @abstractmethod        
    def deposit(self,amount):
        pass

    @abstractmethod
    def withdraw(self,amount):
        pass
```

```python

class SavingAccount(BankAccount):

    def deposit(self,deposit_amount):
        if deposit_amount > 0:
            self.balance += deposit_amount
            return(f"Amount of {deposit_amount} is deposited into your bank account")
        else:
            return("Invalid Amount entered")

    def withdraw(self,withdraw_amount):
        if withdraw_amount:
            self.balance -= withdraw_amount
            return(f"Amount of {withdraw_amount} is withdrawn from your bank account")

        else:
            return("Invalid Amount entered")
```

```python
obj = SavingAccount("Sanjay", 30000)
obj.deposit(300)
```

```python
obj.withdraw(300)
```

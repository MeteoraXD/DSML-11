```python
def precision(TN,FN,FP,TP):
    return TP / (TP + FP)
def accuracy(TN,FN,FP,TP):
    return (TP + TN) / (TP + TN + FP + FN)
def recall(TN,FN,FP,TP):
    return TP / (TP + FN)
def f1_score(p,r):
    return 2 * (p * r) / (p + r)
       
```

```python
p = precision(50, 10, 15, 25)
a = accuracy(50, 10, 15, 25)
r = recall(50, 10, 15, 25)
```

```python
f1 = f1_score(p,r)
```

```python
print(f"Precision is {p:.2f}")
print(f"Accuracy is {a:.2f}")
print(f"Recall is {r:.2f}")
print(f"F1 Score is {f1:.2f}")
```

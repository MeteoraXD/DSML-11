```python
#  all means every element is true
# any means atleast one of the element is true
data_tuple3 = ( 1,2,4,5,43,9,0)
print(len(data_tuple3))
print(max(data_tuple3))
print(any(data_tuple3))
print(min(data_tuple3))
print(sum(data_tuple3))
print(all(data_tuple3))
print(sorted(data_tuple3))
print(tuple(data_tuple3))

```

```python
# unpack
data = ( 1 , 3, 4)
a , b, c = data
print (a )
print ( b )
print ( c)
```

```python
l , b , _ = data
print(l)
print(b)
```

```python
# printing data inside tuple by index
print(data[1])
```

```python
# slicing tuple
print(data[1:3])
```

```python
# Adding 2 tuples
data_1 = (1 , 2)
data_2 = (2 , 3)
print(data_1 + data_2)
```

```python
# Repetitions (Repeating Tuples multiple times using *)
print(data*3)
```

```python
# Membership in tuple (Checking wether an element exists in a tuple)
print (1 in data)
```

```python
# Comparison (comparees tuples elements by elements)
t1 = (1,2,3)
t2 = (1,3,4)
print(t1<t2)
```

```python
# WE CANNOT DELETE INDIVIDUAL ATTRIBUTES OF TUPLE. WE CAN ONLY DO THAT BY DELETING WHOLE OF TUPLE
del t1
```

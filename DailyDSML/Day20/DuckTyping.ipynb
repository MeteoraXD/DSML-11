# Duck Typing

```python
class Duck:
    def fly(self):
        return("Duck is flying")

    def swim(self):
        return("Duck is swiming")    

class Airplane:
    def fly(self):
        return("Airplane flying")

def fly_Test(entity):
    print(entity.fly())
    
```

```python
duck = Duck()
airplane = Airplane()
```

```python
duck.fly()
```

```python
fly_Test(duck)
```

```python
fly_Test(airplane)
```

```python
# Create classes Esewa, khalti and bank transfer, each having a pay(amount) method. Write a payment processor function that works with any payment method object 
class Esewa:
    def pay(self,amount):
        return(f"User has paid {amount} via esewa")

class Khalti:
    def pay(self,amount):
        return(f"User has paid {amount} via khalti")

class BankTransfer:
    def pay(self,amount):
        return(f"User has paid {amount} via Bank transfer")

def payment(entity,amount):
    print(entity.pay(amount))
```

```python
esewa = Esewa()
khalti = Khalti()
bankTransfer = BankTransfer()

```

```python
payment(esewa, 2000)
payment(khalti,3000)
payment(bankTransfer,4000)
```

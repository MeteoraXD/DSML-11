# Magic Functions


```python
class BankCustomer:
    def __init__(self,name,age):
        self.name = name
        self.age = age

# call by print()
def __str__(self):
    return(f"{self.name} is {self.age} years old")

# call by repr()
def __repr__(self):
    return(f"Bank Customer('{self.name}' , '{self.age})")

# called by len()
def __len__(self):
    return self.age

def __call__(self):
    print(f'Age : {self.age}')
```

```python
# create an object
customer = BankCustomer("Sanjay" , 23)

# ____Str____
print(customer)

print(repr(customer))

print(len(customer.age))

print (customer())
```

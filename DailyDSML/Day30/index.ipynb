- Data Manupulation in Pandas

```python
import pandas as pd
titanic_df = pd.read_csv("titanic.csv")
```

```python
titanic_df.plot.box(column= "Age", by="Sex")
```

```python
import  matplotlib.pyplot as plt
titanic_df["Age"].plot.box(
    figsize=(5,6),
    title="Box Plot of Passenger Ages",
    grid=True
)
plt.ylabel("Age")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Only numeric columns will be used for correlation
corr = titanic_df.corr(numeric_only=True)

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap - Titanic Dataset")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

covv = titanic_df.cov(numeric_only=True)

sns.heatmap(covv, annot=True,  cmap="jet")
plt.title("Covariance Heatmap - Titanic Dataset")
plt.show()
```

```python
# Central Tendency

import numpy as np
from scipy import stats
```

```python
x = np.random.standard_t(500,size=1000)
x[:15]
```

```python
_mean = np.mean(x)
print(f"Mean of x is",_mean)
```

```python
plt.hist(x,bins=20,color="blue")
```

```python
plt.axvline(x=_mean,color="yellow")
```

```python
_median = np.median(x)
print(f"Median is:{_median}")
```

```python
plt.hist(x, bins=30, color="white", edgecolor="black")

plt.axvline(x=_median,color="green")
```

```python
import matplotlib.pyplot as plt
from scipy import stats
_mode = stats.mode(x)
print(_mode.mode)

plt.hist(x, bins=30, color="white", edgecolor="black")
plt.axvline(x=_mode.mode, color="black")
plt.show()
```

```python
plt.axvline(x=_mode.mode,color="green")
```

```python
plt.hist(x, bins=30, color="white", edgecolor="black")
plt.axvline(x=_mode.mode, color="black")
plt.axvline(x=_median,color="green")
```

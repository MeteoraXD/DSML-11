# Property

```python
class Student:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self,value):
        if value != "":
            self._name = value
        else :
            raise ValueError("Name cannot be empty")
        self._name = value

```

```python
std = Student("sanjay")
print(std.name)
std.name = "bankai"
print(std.name)
```

```python
class Student1:
    def __init__(self, age,name):
        self._name = name
        self._age = age

    @property
    def age(self):
        return self._age
    @property
    def name(self):
        return self._name

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value
    @age.deleter
    def age(self):
        print(f"Deleting age....{self.age}")
        del self._age
    @name.setter
    def name(self,value):
        if value != "":
            self._name = value
        else :
            raise ValueError("Name cannot be empty")
        self._name = value


std1 = Student1("sanjay",20)
print(std1.name)
print(std1.age)
std1.age = 25
print(std1.age)
std1.name = str(input("Enter your name:"))
print(std1.name)
del std1.age
```

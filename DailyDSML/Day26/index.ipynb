```python
import numpy as np

mat1 = np.array([[1,2 ] ,[3 ,4]])
mat2 = np.array([[1,2 ] ,[3 ,4]])

# matrix addition
mat_add = mat1 + mat2
print("Addition" ,mat_add)

# matrix subtraction
mat_sub = mat1 - mat2
print("Subtraction" ,mat_sub)

# matrix multiplication
mat_mux = mat1 @ mat2
print("Multiplication" ,mat_mux)
```

```python
mat3 =  np.array([[1,2,3]])
mat4 =  np.array([[[1] ,[2],[3]]])

mat5 = mat4 @ mat3
print("shape",mat3.shape )
print("shape",mat4.shape )
print(mat5)
```

```python
# Linear ALgebra in Numpy

# 2 * x + y = 5
# x+ 3y = 6

A = np.array([[2,1] , [1,3]])
B = np.array([5,6])

x = np.linalg.solve(A,B)
x
```

```python
# slope 

x = np.array([[1,1] , [2,1] , [3,1]])
y = np.array([2,3,5])

beta = np.linalg.inv(x.T@x)@x.T@y
print(beta)

print("Slope(m) =" , beta[0])
print("Intercept(C) =" , beta[1])
```

```python
# determinante of matrix

x = np.array([[1,1] , [2,1] ])

determinant = np.linalg.det(x)
print(determinant)
```

```python
# inverse of matrix

y = np.linalg.inv(x).astype(int)
print(y)
```

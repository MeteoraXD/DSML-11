```python
import numpy as np

mat1 = np.array([[1,2 ] ,[3 ,4]])
mat2 = np.array([[1,2 ] ,[3 ,4]])

# matrix addition
mat_add = mat1 + mat2
print("Addition" ,mat_add)

# matrix subtraction
mat_sub = mat1 - mat2
print("Subtraction" ,mat_sub)

# matrix multiplication
mat_mux = mat1 @ mat2
print("Multiplication" ,mat_mux)
```

```python
mat3 =  np.array([[1,2,3]])
mat4 =  np.array([[[1] ,[2],[3]]])

mat5 = mat4 @ mat3
print("shape",mat3.shape )
print("shape",mat4.shape )
print(mat5)
```

```python
m1 =   np.array([[1,2,3]])
m2 =   np.array([[1,2,3]])
m3 = m1.T

m4 = m3 @ m2

m4
```

```python
# matrix division

div = m1 / m2
print("division", div)
```

```python
# scalar matrix

scalar = 2
result = mat1 + 2
print(result)
```

```python
m5 =  np.array([[1,2],[3,4]])
m6 =  np.array([[1],[4]])

m7 = m5 + m6

m7
```

```python
# Linear ALgebra in Numpy

# 2 * x + y = 5
# x+ 3y = 6

A = np.array([[2,1] , [1,3]])
B = np.array([5,6])

x = np.linalg.solve(A,B)
x
```

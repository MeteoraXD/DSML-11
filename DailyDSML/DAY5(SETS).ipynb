```python
# \ is union , & is intersection , - is difference
set_a = {  2 , 4, 6,  8 , 10}
set_b = { 1 , 3 , 5 ,7 , 9}
print(set_a | set_b)
print(set_a & set_b)
print(set_a - set_b)
```

```python
# ^ is symmetric difference 
# in is membership (checks if an element exists in a set)
# <= is subset

print ( set_a ^ set_b)
print ( 3 in set_b)
print ( set_a <= set_b)
```

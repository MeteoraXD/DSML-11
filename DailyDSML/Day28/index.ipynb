````python
import pandas as pd
````

````python
titanic_df.value_counts("Survived")
````

````python
titanic_df.value_counts("Sex")
````

````python
# Normalization of data

titanic_df.value_counts(["Sex", "Survived"])
````

````python
titanic_df.value_counts(["Sex", "Survived"] , normalize= True , ascending=True)
# normalize ley chai 0 - 1 ko range ma value dinxa ( number of count / number of total count)
````

````python
titanic_df[titanic_df["Sex"] == "female"]["Survived"].value_counts(ascending=True , normalize=True)
````

````python
titanic_df[["Age","Fare"]].mean()
````

````python
titanic_df[["Age", "Fare"]].median()
````

````python
# Group by liyera mean nikalni

titanic_df.groupby(['Embarked', 'Survived']).mean(numeric_only=True)
````

````python
# Covariance 
titanic_df[["Age" , "Fare"]].cov()
````

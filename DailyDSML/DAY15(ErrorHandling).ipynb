```python
def divide_numbers(a,b):
    try:
        result = a/b
        print(f"The division between {a} and {b} is {result}")
    except ZeroDivisionError:
        print("You can't divide by zero")


divide_numbers(10,0)
```

```python
try:
    result = 10/0
    print(f"The division is {result}")
except ZeroDivisionError:
    print("You can't divide by zero")
```

```python
try:
    sum = 10 + "20"
    print(f"The sum is {sum}")
except TypeError:
    print("You can't add a number and a string together")
```

```python
# ValueError
try:
    num = int("ten")
    print(f"The number is {num}")
except ValueError:
    print("You can't convert a non-numeric string to an integer")
```

```python
import math
try:
    a= math.sqrt(-1)
    print("The square root is calculated", a)
except ValueError:
    print("You can't take the square root of a negative number")
```

```python
# index error
num = [1,2,3]
try:
    a= num[5]
    print(a)
except IndexError:
    print("Index out of bounds")
```

```python
# key error
my_dict = {"name": "John", "age": 30}
try:
    a = my_dict["gender"]
    print(a)
except KeyError:
    print("Key not found")
```

```python
# file not found error
try:
    with open("index.txt", "r") as file:
        content = file.read()
        print('file content:', content)
except FileNotFoundError:
    print("File not found")
finally:
    print("This block will always be executed")    
```

```python
# file upload using try catch finally

try:
    with open("index.txt", "r") as file:
        content = file.read
        print('File Contnet' , content)
except:
    print('File not found')
finally:
    print('Process completed')            
```

```python
def check(a):
    if a >= 0 :
        print('positive')
    else:
        raise ValueError('Number cannot be negative')
check(-7)            
```

```python
try:
    check(-4)
except ValueError as e :
    print(e)    
```

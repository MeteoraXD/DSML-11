**LDA** 
- LDA is supervised learning primarily used to dimessionality reduction and classification tasks
- The goal is to find a linear combination of features that maximize the seperation of class while minnimizing the spread (variance) within each class.
        - maximizing mean difference
        - minimizing within class scatter

Steps for LDA
 - Calculate the Mean vectors for each class
 - compute the covariance of each class
 - compute the within class scatter matrix
 - compute the between class scatter matrix
 - find eigen vlaues and choose top eigen value
 - find the eigen vectors
 - find the dot product of data points with eigen vector

        

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df = sns.load_dataset('titanic') 
```

```python
df.columns
```

```python
X = df[['pclass', 'age', 'sibsp', 'parch', 'fare']]
```

```python
scalar = StandardScaler()
X = X.fillna(X.mean())
X_Scaled = scalar.fit_transform(X)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(X_Scaled)
```

```python
pca_df = pd.DataFrame(data=x_pca, columns=['PC1', 'PC2'])
pca_df.head()
```

```python
pca_2 = PCA(n_components=3)
```

```python
x_pca_1 = pca_2.fit_transform(X_Scaled)
```

```python
pca_df_1 = pd.DataFrame(data=x_pca_1, columns=['PC1', 'PC2', 'PC3'])
pca_df_1.head()
```

```python
# Applying LDA

from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
y= df['survived']

lda = LinearDiscriminantAnalysis(n_components=1)
x_lda = lda.fit_transform(X_Scaled, y)
```

```python
lda_df = pd.DataFrame(data=x_lda, columns=['LD1'])
```

```python
lda_df.head()
```

 ** Load Iris dataset and perform PCA**
 - Standardize the features.
 - Reduce the dataset from 4D to 2 principal components.
 - Calculate the explained variance ratio.
 - Plot the transformed data using different colors for each species.



```python
df_iris = sns.load_dataset('iris') 
```

```python
df_iris.head()
```

```python
X_Iris = df_iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
```

```python
scalar = StandardScaler()
X_Iris = X_Iris.fillna(X_Iris.mean())
X_Iris_Scaled = scalar.fit_transform(X_Iris)
```

```python
pca = PCA(n_components=2)
x_iris_pca = pca.fit_transform(X_Iris_Scaled)
```

```python
pca_df = pd.DataFrame(data=x_iris_pca, columns=['PC1', 'PC2'])
pca_df.head()
```

```python
y = df_iris['species']

lda = LinearDiscriminantAnalysis(n_components=2)
x_lda = lda.fit_transform(X_Iris_Scaled, y)
```

```python
lda_df = pd.DataFrame(data=x_lda, columns=['LD1', 'LD2'])
lda_df.head()
```

```python
import matplotlib.pyplot as plt

lda_df['species'] = df_iris['species'].values

plt.figure(figsize=(8, 6))
sns.scatterplot(data=lda_df, x='LD1', y='LD2', s=60)
plt.xlabel('LD1')
plt.ylabel('LD2')
plt.title('LDA: Species Projection')
plt.show()
```

```python
print("Explained variance ratio (first two components):", pca.explained_variance_ratio_)
```

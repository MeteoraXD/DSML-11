```python
import pandas as pd
students_df = pd.read_csv("survey_student.csv")
```

```python
students_df.columns =["Date","Gender","Grade","Year","Faculty","Average","Accomodation","Monthly_Allowance","Scholarship_Bulsary","Study_Hours","Party_Socializing","Drinks_Consumed","Class_Missed","Modules_Failed","Romantic_Relationship","Alcohol_Consumption_Allowed","Relationship_Parents"]


print(students_df.shape)
print(students_df.info())
print(students_df.isnull().sum())
```

```python
students_df.drop(columns = ["Party_Socializing","Drinks_Consumed","Romantic_Relationship","Alcohol_Consumption_Allowed","Relationship_Parents"],inplace = True)
```

```python
students_df.Gender.fillna(students_df["Gender"].mode()[0],inplace = True)
```

```python
students_df.Faculty.fillna(students_df["Faculty"].mode()[0],inplace = True)
```

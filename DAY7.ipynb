```python
marks = float(input("Enter your GPA: "))

if 0 <= marks <= 4:
    if marks >= 3.6:
        print("Your grade is A+")
    elif marks >= 3.2:
        print("Your grade is A")
    elif marks >= 2.8:
        print("Your grade is B+")
    elif marks >= 2.4:
        print("Your grade is B")
    elif marks >= 2.0:
        print("Your grade is C+")
    elif marks >= 1.8:
        print("Your grade is C")
    elif marks >= 1.6:
        print("Your grade is D")
    else:
        print("You have failed")
else:
    print("Invalid CGPA provided")        
```

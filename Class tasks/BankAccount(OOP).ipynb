Create a Bank Account class and implement `__str__()` and `__repr__()` methods to display account details in user-friendly and deverloper-friendly formats

```python
class BankAccount:
    def __init__(self,name,account_number):
        self.name = name
        self.account_number = account_number

    def __str__(self):
        return(f"Account name : {self.name} and Account number : {self.account_number}")

    def __repr__(self):
        return(f"Account Details : ('{self.name}' , '{self.account_number}')")
```

```python
account_Details = BankAccount('Sanjay', 1234353434)
```

```python

print(account_Details)
```

```python

print(repr(account_Details))
```

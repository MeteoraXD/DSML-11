```python
# ATM system using error handling

# Develop a mini atm program that uses custom exceptions, try-except and finally to handle deposits , withdrawls and balance checks safely

balance = 52000.0

while True:
    print("\n=============================")
    print("      Welcome to Our ATM     ")
    print("=============================")
    print("Enter 1 for Balance Check")
    print("Enter 2 for Deposit")
    print("Enter 3 for Withdrawal")
    print("Enter 4 to Exit")
    print("-----------------------------")

    try:
        user_choice = int(input("Enter your choice: "))

        if user_choice == 1:
            print(f"Your current balance is: Rs.{balance}")

        elif user_choice == 2:
            deposit_amount = float(input("Enter amount to deposit: Rs."))
            if deposit_amount <= 0:
                raise ValueError("Deposit amount must be greater than zero.")
            balance += deposit_amount
            print(f"Deposit successful! New balance: Rs.{balance}")

        elif user_choice == 3:
            withdraw_amount = float(input("Enter amount to withdraw: Rs."))
            if withdraw_amount <= 0:
                raise ValueError("Withdrawal amount must be greater than zero.")
            if withdraw_amount > balance:
                raise ValueError(f"Insufficient funds. Available balance: Rs.{balance}")
            balance -= withdraw_amount
            print(f"Withdrawal successful! New balance: Rs.{balance}")

        elif user_choice == 4:
            print("\nThank you for using our ATM. Goodbye!")
            break

        else:
            raise ValueError("Invalid choice. Please enter 1, 2, 3, or 4.")

    except ValueError as e:
        print(f"ERROR: {e}")

    finally:
        print("Thank you for using our bank.")
```

Create a Polygon class and use it to represent a triangle, square , Pentagon and Hexagon


```python
class Polygon:
    def triangle(self):
        print('I am triangle')

    def square(self):
        print('I am square')
    def Rectangle(self):
        print('I am Rectangle')
    def Pentagon(self):
        print('I am a Pentagon')
    def Hexagon(self):
        print('I am a Hexagon')


```

```python
obj = Polygon()
obj.Hexagon()
```

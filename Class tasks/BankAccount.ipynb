```python
class BankAccount:
    def __init__(self, name, bank_balance=0):
        self.name = name
        self.__bank_balance = bank_balance

    def create_account(self):
        print(f"Bank Account Created for {self.name} with balance {self.__bank_balance}")

    def deposit(self, amount):
        if amount > 0:
            self.__bank_balance += amount
            print(f"Added {amount} to Account")

    def withdraw(self, amount):
        if amount <= 0:
            print("Insufficient amount to withdraw")
        elif amount > self.__bank_balance:
            print("Insufficient balance")
        else:
            self.__bank_balance -= amount
            print(f"Deducted {amount} from your account")

    def check_balance(self):
        print(f"{self.name} has balance of {self.__bank_balance}")
        return self.__bank_balance
```

```python
class SavingAccount(BankAccount):
    def add_interest(self, interest_rate=5):
        balance = self.check_balance()
        interest = (balance * interest_rate) / 100
        print(f"Account interest: {interest}")
        return interest

```

```python
print("--- Enter Account Details ---")
name = str(input("Enter your name"))
amount = int(input("Enter your amount"))

obj = BankAccount(name,amount)
obj.create_account
obj.deposit(600)

```

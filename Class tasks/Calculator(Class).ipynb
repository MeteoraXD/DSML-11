# Calculator using Class and object

```python
class Calculator:
    company_name = 'XYZ'

    def __init__(self):
        self.num1 = int(input('Enter first number: '))
        self.num2 = int(input('Enter second number: '))
        self.operator = input('Enter an operator (+, -, *, /): ')

    def calc(self):
        if self.operator == '+':
            return self.num1 + self.num2
        elif self.operator == '-':
            return self.num1 - self.num2
        elif self.operator == '*':
            return self.num1 * self.num2
        elif self.operator == '/':
            return self.num1 / self.num2
        else:
            return 'Invalid Operator'

    @classmethod
    def get_company_name(cls):
        return cls.company_name

    @staticmethod
    def welcome_message():
        return 'Welcome to our company'

calculator = Calculator()
print(calculator.calc())
print(Calculator.get_company_name())
print(Calculator.welcome_message())
```

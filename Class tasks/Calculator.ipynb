# CALCULATOR TASK


```python
print("+ for addition:")
print("- for subtraction")
print("/ for Division")
print("* for Multiplication")

c = input("What operation would you like to enter: ")


def calculator(a,b,c):
    if c == '+':
        return a+b
    elif c == '-':
        return a-b    
    elif c == '/':
        return a/b
    elif c == '*':
        return a*b   
             

a = float(input('Enter a number'))
b = float(input('Enter a number'))

print (calculator(a,b,c))
```
